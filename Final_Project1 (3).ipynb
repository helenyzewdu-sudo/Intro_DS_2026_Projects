#### Key analytical features:

```python
print(f'Missing values within the original data set:')
crime_incidents.isnull().sum()
```

When checking for missing values within the original data set, multiple variables showed missing data. For example, the variable END_DAT, DISTRICT, PSA, and BID holds a small number of missing values; however, the variable OCTO_RECORD_ID is completely useless as the entire data values are missing. Fortunately, these variables are not important for my desired research analysis. 

```python
print(f'Checking for duplicated records:')
crime_incidents.duplicated().sum()

crime_incidents = crime_incidents.drop_duplicates()

print(crime_incidents.duplicated().sum())
```

When checking for duplicated values, there were no rows that were duplicated within my data set. This analysis shows that each value or crime shown is a unique and distinctive crime that has occurred. Due to there being no duplicated, no values or columns wer removed. 

```python
print(f'Standardized column names:')
crime_incidents.columns = crime_incidents.columns.str.lower().str.strip().str.replace(" ", "_")
crime_incidents.columns
```

Standardizing the column names within my data converted the column names to be consistent with each other. Doing this made each column name lower case as well as removed unnessecary spaces that would have resulted in errors. Standardizing these columns makes the data set appear cleaner. 

```python
print(f'Standardized string entities:')
text_columns = crime_incidents.select_dtypes(include="object").columns

for col in text_columns:
    crime_incidents[col] = crime_incidents[col].str.strip().str.lower()
    
crime_incidents.columns
```

Standardizing the column names within my data converted the column names to be consistent with each other. Doing this made each column name lower case as well as removed unnessecary spaces that would have resulted in errors. Standardizing these columns makes the data set appear cleaner. 

```python
print(f'Parsing date strings to datetime objects:')
crime_incidents['report_dat'] = pd.to_datetime(crime_incidents['report_dat'], errors='coerce')
crime_incidents['start_date'] = pd.to_datetime(crime_incidents['start_date'], errors='coerce')
crime_incidents['end_date'] = pd.to_datetime(crime_incidents['end_date'], errors='coerce')
crime_incidents['hour'] = crime_incidents['start_date'].dt.hour

crime_incidents.info()
```

Parsing date strings to date time objects is important to improve the clarity and ensure time-based analysis. When analyzing the values within my data, it is important to ensure the exact time when looking at the values because they were originally strings. For example, I converted "2025-05-19 03:24:00" , which is a string, to a datetime that python will understand and interpret, datetime64. Coming from the start_dat column, I created a new variable called "hour" to ensure that the specific time is analyzed and supports my research question. 

```python
print(f'Encoded important categorical variables:')
crime_incidents['offense_encoded'] = crime_incidents['offense'].astype('category').cat.codes

crime_incidents['offense_encoded']
```

I encoded the categorical variable "offense" to numerical in order for numerical analysis to be efficient and for the data visualization to exhibit clear visuaizations. The numerical values do not represent an improtant meaning such as one being more significant than the other, but rather incorporates the integration of categorical values in statistical analysis. 

```python
print(f'Frequency of each crimes and the encoded offense:')
crime_incidents['offense_encoded'].value_counts()

crime_incidents[['offense', 'offense_encoded']].drop_duplicates()
```

```python
print(f'Irrelevant columns dropped.') 
crime_incidents = crime_incidents.drop(
    columns=['x', 'y', 'ccn', 'objectid', 'octo_record_id', 'xblock', 'yblock', 'district', 'psa', 'bid', 'anc', 'ward', 'census_tract', 'block_group'],
    errors='ignore'
)
```

When looking at this large data set of crime, there were many unnecessay columns/variables that were irrelevant to my analysis. In order for data analysis to be more clear and efficient. These columns were dropped. 

```python
crime_incidents.to_csv("cleaned_crime_data.csv", index=False)
```

# Data Visualization and Interpretation

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

#### Histogram
The purpose of this intent when using this histogram is to analyze the frequency of crimes that occur during each hour. 

```python
sns.histplot(crime_incidents_filtered['hour'], bins=24, color='skyblue', kde=True)

plt.title("Distribution of Crime Occurrences by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Crimes")

plt.xticks(range(0, 24), rotation=45)

plt.savefig('crime_hist_hour.jpg')
plt.show()
```

#### Interpretation: 
Thiis univariate graph, the histogram exhibits that crimes not stay consistent throughout the day but rather decreas/increase throughout the hours. Hours 15 to 21 show higher frequencies, suggesting that more crimes occur during the evening and night time. The histogram shows that crimes are not stable throughout the hours of the day.

#### Scatterplot
The purpose of intent when using this scatterplot is to analyze the relationship between the latitude of a crime (geographic location) and the hour in which the crime occurs. It also analyzes how the crimes scatter when looking at the different shift or times of the day (morning, afternoon, evening, night.)

```python
sns.scatterplot(
    data=crime_incidents_filtered,
    y='latitude',
    x='hour',
    hue='shift',
    palette='viridis'
)

plt.title('Crime by hour and location')
plt.xlabel('Hour')
plt.ylabel('Location')

plt.savefig('crime_scatter_hour_location.jpg')
plt.show()
```

#### Interpretation: 
This scatterplot shows the relationship between the geographic location of crimes and the time of day in which they occur. The latitude ranging from 38.825-39.00suggest that the location of crimes occuring are in the same or similar geographic location (Washington, D.C, as mentioned in the article.) Though the scatterplot is exhibit consistent crimes occurring different different hours of the day, there does not seem to be a strong influence of latitude on shift or hour of the crime committed. 

#### Heat map
The purpose of intent in using this graph is to analyze the strength and relationship between my variables to better understand the correlation between them. 

```python
plt.figure(figsize=(8,6))

corr_matrix = crime_incidents_filtered.corr(numeric_only=True)

ax = sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.3f',
    cmap='magma',
    center=0,
    linewidth=0.5,
    cbar_kws={'label': 'Correlation'}
)

plt.title('Correlation Heat Map of Numeric Crime Features')

plt.savefig('Crime_Correlation_Heatmap.jpg')
plt.show()
```

#### Interpretation
When analyzing the heat map, it is evident that there are weak relationships (numbers closer to zero) between my variables, indicating no strong relationship between the variables, suggesting that crimes happening may be affected by other qualitative variables. 

#### Animated Line Chart 
The purpose of intent in using this graph is to provide an animated visual analyzation of the trends of crime occurrences and how they change over the hour of the day. It also shows how the different hours slides over different crimes. 

```python
df_anim = crime_incidents_filtered.groupby(['hour', 'shift', 'offense']).size().reset_index(name='crime_count')
```

```python
fig = px.line(
    df_anim,
    x='hour',
    y='crime_count',
    color='shift',
    animation_frame='offense'
)

fig.show()
```

#### Interpretation
This animated line plot chart allows the analyses and interaction of the different varaiables and how they shift during different hours of the day, dependent on the offense that has occured. 

```python
# geomap filter for evening crimes
crime_geo = crime_incidents_filtered[crime_incidents_filtered['shift'] == 'evening']

crime_geo = crime_geo.rename(columns={
    'latitude': 'Lat',
    'longitude': 'Lon',
    'offense': 'Crime_Type'
})

crime_geo
```

#### Geo map
The purpose of intent in using this graph is to provide another animated visualization showing the location of which crimes occur during the evening. 

```python
fig = px.scatter_mapbox(
    crime_geo,
    lat='Lat',
    lon='Lon',
    color='Crime_Type',
    hover_name='Crime_Type',
    zoom=10,
    height=600,
    title='Interactive Map of Evening Crime Incidents'
)

fig.update_layout(mapbox_style='open-street-map')
fig.show()
```

#### Interpretation
When analyzing this graph, it is evident that crimes occur within the same city due to the visual cluster as well as the close and approixmate latitude and longitudinal values. It is evident that majority of crimes occur in the central part of Washington, D.C. As seen in the visualization, the blue clustered dots represent a high frequency of robberies occuring in the center of Washington, D.C.

### ViolinPlot
The purpose of intent in using this graph is to show a unique visualization of the spread of crimes and their frequency across different times 

```python
sns.violinplot(
    data=crime_incidents_filtered,
    x='shift',
    y='hour',
    hue='shift',
    palette=['pink', 'red', 'purple'],
    inner='box',
    fill=True
)

plt.title('Crime hours by shift')
plt.xlabel('Shift')
plt.ylabel('Hour of the Day')
plt.xticks(rotation=30)

plt.savefig('crime_violin_shift_hour.jpg')
plt.show()
```

#### Interpretation
Similar to the other interpretations of the graph, this unique visualization displays  the crimes and their variation when shifting across the times of day. Within the violin representing the midnight shift, it is evident that majority crimes occur during the first few hours (late night, early morning). Within the day shift, it is evident that most crimes that occur are moderately spread, and during the evening, majority of the crimes occur during the hours of 16-22. These violin-like visualizations show the periods of highest frequency. 

# Additional Findings/Analysis

Additional Research Question: Does crime frequency differ by neighborhood cluster? 

#### Count Plot
The purpose of intent when analyzing this visualization is to understand the pattern of occurence of crimes within different neighborhood clusters. 

```python
sns.countplot(
    data=crime_incidents_filtered,
    y='neighborhood_cluster',
    order=crime_incidents_filtered['neighborhood_cluster'].value_counts().index,
)

plt.title('Crime Frequency by Neighborhood Cluster')
plt.xlabel('Crimes committed')
plt.ylabel('Neighborhood Cluster')

plt.savefig('crime_by_neighborhood.jpg')
plt.show()
```

#### Interpretation
This visualization exhibits that crimes committed exhibits notable variation between clusters. While the neighborhood cluster 41 has nearly 0 crimes, cluster 8 and 25 contribute to the higher crime rate in Washington D.C. This goes into depth with the geographic location, showing that other factors such as neighborhood plays a significant factor into crime frequency. 

Additional research question: Does latitude (geographic location) vary by the type of offense committed? 

#### Boxplot
The purpose of intent for this visualization is to analyze wether the type of crime differes within the specified neighborhood. 

```python
sns.boxplot(
    data=crime_incidents_filtered,
    x='offense',
    y='latitude'
)

plt.title('crime by location')
plt.xlabel('Offense Type')
plt.ylabel('Latitude')
plt.xticks(rotation=45)

plt.savefig('boxplot_latitude_offense.jpg')
plt.show()
```

#### Interpretation 
This visualization shows that the crime types that occur are in the same geographic areas with minor fluctuations. While this plot shows the mean and median of the variables we are analyzing, it shows that the median latitude does not significantly differ in relation to the type of crime occurred. As seen within the boxplot, the points are overlapping, suggesting that the locations do not have high variablity. There are outiers within this visualization, suggesting that some crimes do occur outside of the greater proportion.
